Vamos utilizar o dataset de liberdade econômica criado pelo Heritage Foundation.

O Índice de Liberdade Econômica (Index of Economic Freedom) é uma classificação que avalia o grau de liberdade econômica de 186 países. Criado em 1995 através de uma parceria entre o The Wall Street Journal e o think tank conservador norte-americano conhecido como Heritage Foundation, o índice considera doze categorias de "liberdade econômica": nos negócios; no comércio; liberdade fiscal; de intervenção do governo; monetária; de investimentos; financeira; de corrupção; do trabalho; e direitos de propriedade [8].

Vamos trabalhar com **12 features**. Os países são avaliados em quatro categorias principais, e cada uma contém três subcategorias:
* **Estado de Direito**: Direitos de Propriedade, Integridade de Governo, Eficiência Judicial
* **Tamanho do governo**: Gastos do Governo, Carga Tributária, Saúde Fiscal
* **Eficiência Regulatória**: Liberdade Comercial, Liberdade de Trabalho, Liberdade Monetária
* **Mercados Abertos**: Liberdade de Comércio Exterior, Liberdade de Investimento, Liberdade Financeira.

Todos os critérios têm o mesmo peso no cálculo final do índice, que considera uma **escala de 0 a 100**. Com base nesse resultado, a Heritage divide os países em cinco categorias principais: 
* **Livre** (entre 80 e 100)
* **Majoritariamente livre** (entre 70 a 79,9)
* **Moderadamente livre** (60 a 69,9)
* **Majoritariamente não-livre** (entre 50 a 59,9)
* **Reprimido** (0 a 49,9)
* **Não ranqueado**

##### Exercício 1 - Hiperparâmetros

Vamos verificar quais as dimensões dos nossos dados de entrada e saída? Quais os hiperparâmetros devemos definir para o treinamento deste modelo? 

```python
# Instanciando o algoritmo
# eu entendi que o tamanho do grid e proporcional ao tamanho do dataset, 
# sem levar em consideracao o numero de features, porque cada neuronio tera um peso para cada feature
# como o dataset tem 186 paises, vi que uma forma ideal eh tirar a raiz quadrada de 186, que eh 13.638
# com um grid de 4x4 eu tenho 16 neuronios, o que acredito ser suficiente

grid_width = 4
grid_height = 4

sofm = MiniSom(
    x = grid_width,
    y = grid_height,
    learning_rate=0.1, # deixei como 0.1, para iniciar
    input_len=12,
)
```

#### Exercício 2

O objetivo deste exercício é verificar como o número de iterações influência no treinamento do modelo. Além disso, é interessante verificar como os hiperparâmetros podem ser modificados e o impacto deles no resultado.

Faça outros experimentos com o número de neurônios diferente. Lembre-se que quanto maior o número de neurônios, maior a capacidade de representação.

```python
# Executando o treinamento
sofm.train(data=data_train, num_iteration=10)
```

Avaliação do modelo. Para o escopo deste exercício o avaliação pode ser feita de forma qualitativa.

```python
# Verificando os pesos dos primeiros neurônios
sofm.get_weights().shape
```

```python
def plot_u_matrix(sofm):
    plt.figure(figsize=(6, 5))
    plt.pcolor(sofm.distance_map().T, cmap='bone_r')
    plt.colorbar()
    plt.title('Mapa de Distâncias')
    plt.show()

plot_u_matrix(sofm)
```

```python
def count_neurons_used(sofm, data):
    winners = []
    for i in range(len(data)):
        winners.append(sofm.winner(data[i]))

    unique_winners = len(set(winners))
    total_neurons = sofm._weights.shape[0] * sofm._weights.shape[1]

    print(f"Neurônios utilizados: {unique_winners}/{total_neurons}")
    print(f"Porcentagem de uso: {(unique_winners/total_neurons)*100:.1f}%")

```

```python
count_neurons_used(sofm, data_train)
```

```python
# eu entendi que tem 2 formas de avaliar o modelo, plotando esse mapa de distancias que mostra os clusters dos neuronios, 
# ou seja, o quanto o mapa convergiu, e o quanto o mapa esta separando os clusters,
# e quantos neuronios foram utilizados, o que mostra se o modelo sofreu underfitting ou overfitting

# no caso o modelo claramente sofreu underfitting, entao vou treinar com mais iteracoes

def train_sofm(data_train, grid_width, grid_height, iterations, learning_rate=0.1):
    sofm = MiniSom(
        x = grid_width,
        y = grid_height,
        learning_rate=0.1,
        input_len=12,
    )

    sofm.train(data_train, iterations)


    plot_u_matrix(sofm)
    count_neurons_used(sofm, data_train)
    return sofm
```

```python
sofm = train_sofm(
    data_train, 
    grid_width, 
    grid_height, 
    100,
)
```

```python
# aqui ja teve uma melhora, mas talvez seja melhor mais iteracoes
sofm = train_sofm(
    data_train, 
    grid_width, 
    grid_height, 
    1000,
)
```

```python
# Aqui parece que o modelo teve um bom resultado, 
# com uma distribuicao clara no mapa de distancias 
# e com uma boa porcentagem de neuronios usados.
```

Podemos analisar também quão bem um determinado neurônio foi representado, comparando o vetor de pesos do neurônio e o padrão de entrada da rede.

Faça o predict para um subconjunto de países e verifique os neurônios aos quais ele foi mapeado.

```python
countries = ['Brazil']

# Selecionando as features (padrão de entrada dos países escolhidos acima)
list_country = []
for item in countries:
    country_data = data[features][data['Country Name']==item].values
    if len(country_data) > 0:
        list_country.append(sofm.winner(country_data[0]))

print(list_country)
for i, val in enumerate(list_country):
    print(f'{countries[i]} -> neuron number: {str(list_country[i])}')
```

Quão bem um determinado país foi representado? Verifique os dados de entrada e os pesos associados ao neurônio correspondente

```python
def verificar_representacao(pais, sofm, data, features):
    dados_pais = data[features][data['Country Name'] == pais].values[0]
    
    posicao_neuronio = sofm.winner(dados_pais)
    pesos_neuronio = sofm.get_weights()[posicao_neuronio[0], posicao_neuronio[1]]
    
    distancia = np.linalg.norm(dados_pais - pesos_neuronio)
    
    print(f"País: {pais}")
    print(f"Neurônio: {posicao_neuronio}")
    print(f"Distância: {distancia:.2f}")
```

```python
# Apos pesquisar novamente, vi que o ideal seria um distancia abaixo de 10, vou tentar retreinar com mais neuronios.
verificar_representacao('Brazil', sofm, data, features)
```

Quais países foram atribuídos à um mesmo neurônio?

```python
def paises_por_neuronio(sofm, data, features):
    neuronios = {}
    for i, pais in enumerate(data['Country Name']):
        dados_pais = data[features].iloc[i].values
        posicao = sofm.winner(dados_pais)
        if posicao not in neuronios:
            neuronios[posicao] = []
        neuronios[posicao].append(pais)
    return neuronios
grupos = paises_por_neuronio(sofm, data, features)
neuronios = list(grupos.keys())
neuronios[0], len(neuronios)
```

#### Exercício 4 - Visualização dos dados/resultados

A visualização dos dados pode ser feita em um grid bidimensional, independente das dimensões dos dados de entrada. Neste caso, os dados de entrada para treinamento do modelo tem 12 dimensões (12 features selecionadas no dataset). 

Quais observações vocês podem fazer com relação aos dados apresentados no gráfico? Alterariam algum hiperparâmetro? Por que?

```python
def analisar_neuronios(neuronios_escolhidos, grupos):
    resultados = {}
    
    for neuronio in neuronios_escolhidos:
        if neuronio in grupos:
            paises = grupos[neuronio]
            
            dados_neuronio = data[data['Country Name'].isin(paises)]
            
            medias = dados_neuronio[features].mean().values
            
            regioes = dados_neuronio['Region'].value_counts().to_dict()
            
            resultados[neuronio] = {
                'paises': paises,
                'medias': medias,
                'regioes': regioes,
                'qtd_paises': len(paises)
            }
    

    fig, ax = plt.subplots(figsize=(12, 6))
    
    x = np.arange(len(features))
    width = 0.8 / len(neuronios_escolhidos)
    
    for i, neuronio in enumerate(neuronios_escolhidos):
        if neuronio in resultados:
            offset = (i - len(neuronios_escolhidos)/2) * width + width/2
            ax.bar(x + offset, resultados[neuronio]['medias'], width, 
                   label=f'Neurônio {neuronio}')
    
    ax.set_xlabel('Features')
    ax.set_ylabel('Média')
    ax.set_title('Comparação de Médias por Neurônio')
    ax.set_xticks(x)
    ax.set_xticklabels(features, rotation=45, ha='right')
    ax.legend()
    plt.tight_layout()
    plt.show()
    
    for neuronio, info in resultados.items():
        print(f"Neurônio {neuronio}:")
        print(f"  Países: {info['paises']}")
        print(f"  Regiões: {info['regioes']}")
        print()
```

```python
resultado = analisar_neuronios([(5,5), (7,0)],grupos)
```

```python
resultado = analisar_neuronios([(5,5), (6,6), (5,6)],grupos)
```

```python
resultado = analisar_neuronios([(7,1), (7,0)],grupos)
resultado = analisar_neuronios([(0,7), (1,7)],grupos)
```
